## 🔴 Ejercicio 7: Combinación de DataFrames (Merge)

Tienes dos DataFrames: uno con información de pedidos y otro con información de clientes.

```python
pedidos = pd.DataFrame({
    'PedidoID': [101, 102, 103, 104, 105],
    'ClienteID': [1, 2, 1, 3, 2],
    'Monto': [250.0, 180.5, 320.0, 95.0, 410.0],
    'Fecha': pd.to_datetime(['2025-01-15', '2025-01-16', '2025-01-16', '2025-01-17', '2025-01-18'])
})

clientes = pd.DataFrame({
    'ClienteID': [1, 2, 3, 4],
    'Nombre': ['Ana García', 'Carlos Ruiz', 'María López', 'Juan Pérez'],
    'Ciudad': ['Madrid', 'Barcelona', 'Valencia', 'Sevilla']
})
```

Realiza:
1. Un merge (inner join) entre pedidos y clientes usando 'ClienteID'
2. Muestra el resultado ordenado por 'Monto' descendente
3. Calcula el monto total de pedidos por ciudad
4. Identifica el cliente que más ha gastado (total de sus pedidos)

````python
# Título: combinación de Dataframes (Merge)
# Autor: Ivana Sánchez Pérez

# colores
rojo = '\033[91m'
verde = '\033[92m'
azul = '\033[94m'
magenta = '\033[95m'
amarillo = '\033[93m'
rosa = '\033[38;5;200m'
turquesa = '\033[38;5;44m'
azul_marino = '\33[38;5;67m'
lima = '\33[38;5;46m'
reset = '\033[0m'

import pandas as pd

# Crear los DataFrames
pedidos = pd.DataFrame({
    'PedidoID': [101, 102, 103, 104, 105],
    'ClienteID': [1, 2, 1, 3, 2],
    'Monto': [250.0, 180.5, 320.0, 95.0, 410.0],
    'Fecha': pd.to_datetime(['2025-01-15', '2025-01-16', '2025-01-16', '2025-01-17', '2025-01-18'])
})

clientes = pd.DataFrame({
    'ClienteID': [1, 2, 3, 4],
    'Nombre': ['Ana García', 'Carlos Ruiz', 'María López', 'Juan Pérez'],
    'Ciudad': ['Madrid', 'Barcelona', 'Valencia', 'Sevilla']
})

print(f"{azul}DataFrame Pedidos{reset}")
print(f"{turquesa}{pedidos}{reset}")
print(f"\n{azul}DataFrame Clientes{reset}")
print(f"{turquesa}{clientes}{reset}")

# Merge (INNER JOIN) entre pedidos y clientes

pedidos_clientes = pd.merge(pedidos, clientes, on='ClienteID', how='inner')

print(f"\n{verde}Resultado del merge (INNER JOIN){reset}")
print(f"{turquesa}{pedidos_clientes}{reset}")


# Ordenar por monto descendente

pedidos_ordenados = pedidos_clientes.sort_values(by='Monto', ascending=False)

print(f"\n{magenta}Pedidos ordenados por monto (descendente){reset}")
print(f"{turquesa}{pedidos_ordenados}{reset}")

# Monto total de pedidos por ciudad
monto_por_ciudad = pedidos_clientes.groupby('Ciudad')['Monto'].sum()

print(f"\n{turquesa}Monto total de pedidos por ciudad{reset}")
print(f"{magenta}{monto_por_ciudad}{reset}")

# Cliente que más ha gastado

gasto_por_cliente = pedidos_clientes.groupby('Nombre')['Monto'].sum()
cliente_top = gasto_por_cliente.idxmax()
monto_max = gasto_por_cliente.max()

print(f"\n{rosa}Cliente que más ha gastado{reset}")
print(f"{lima}{cliente_top} con un total de {monto_max:.2f} €{reset}")
````

## 🔴 Ejercicio 8: Análisis de Series Temporales

Crea un DataFrame con datos de temperatura diaria durante enero 2025:

```python
np.random.seed(42)
fechas = pd.date_range('2025-01-01', '2025-01-31', freq='D')
temperaturas = 10 + 5 * np.sin(np.arange(31) * 2 * np.pi / 31) + np.random.normal(0, 2, 31)

df_temp = pd.DataFrame({
    'Fecha': fechas,
    'Temperatura': temperaturas
})
df_temp.set_index('Fecha', inplace=True)
```

Realiza:
1. Calcula la media de temperatura del mes
2. Encuentra los 5 días más cálidos
3. Calcula la temperatura promedio por semana (`.resample('W').mean()`)
4. Cuenta cuántos días la temperatura estuvo por encima de 12°C

````python
# Título: Análisis de Series Temporales
# Autor: Ivana Sánchez Pérez

# colores
rojo = '\033[91m'
verde = '\033[92m'
azul = '\033[94m'
magenta = '\033[95m'
amarillo = '\033[93m'
rosa = '\033[38;5;200m'
turquesa = '\033[38;5;44m'
azul_marino = '\33[38;5;67m'
lima = '\33[38;5;46m'
reset = '\033[0m'

import pandas as pd
import numpy as np

# Crear el DataFrame
np.random.seed(42)
fechas = pd.date_range('2025-01-01', '2025-01-31', freq='D')
temperaturas = 10 + 5 * np.sin(np.arange(31) * 2 * np.pi / 31) + np.random.normal(0, 2, 31)

df_temp = pd.DataFrame({
    'Fecha': fechas,
    'Temperatura': temperaturas
})
df_temp.set_index('Fecha', inplace=True)

print(f"{azul}DataFrame de temperaturas (Enero 2025):\n{reset}")
print(f"{turquesa}{df_temp}{reset}")

# 1. Media de temperatura del mes
media_mes = df_temp['Temperatura'].mean()
print(f"\n{lima}Media de temperatura del mes: {media_mes:.2f} °C {reset}")

# 2. Los 5 días más cálidos
dias_calidos = df_temp.nlargest(5, 'Temperatura')
print(f"\n{azul}Los 5 días más cálidos {reset}")
print(f"{turquesa}{dias_calidos}{reset}")

# 3. Temperatura promedio por semana
promedio_semanal = df_temp.resample('W').mean()
print(f"\n{azul}Temperatura promedio por semana{reset}")
print(f"{turquesa}{promedio_semanal}{reset}")

# 4. Días con temperatura mayor a 12°C
dias_mayor_12 = (df_temp['Temperatura'] > 12).sum()
print(f"\n{lima}Días con temperatura mayor a 12°C: {dias_mayor_12}{reset}")

````

````python
# Título: Limpieza y Transformación de Datos
# Autor: Ivana Sánchez Pérez

# colores
rojo = '\033[91m'
verde = '\033[92m'
azul = '\033[94m'
magenta = '\033[95m'
amarillo = '\033[93m'
rosa = '\033[38;5;200m'
turquesa = '\033[38;5;44m'
azul_marino = '\33[38;5;67m'
lima = '\33[38;5;46m'
reset = '\033[0m'

import pandas as pd

# Crear el DataFrame
datos_sucios = pd.DataFrame({
    'producto': ['  laptop  ', 'MOUSE', 'Teclado', 'laptop', 'mouse', '  Monitor'],
    'precio': ['1200€', '25', '45.50€', '1150', '30€', '350'],
    'stock': ['10', '  50  ', '30', '5', 'agotado', '15'],
    'categoria': ['Informática', 'informática', 'Informática', 'Informática', 'Informática', 'pantallas']
})

print(f"\n{azul}DataFrame original{reset}")
print(f"{turquesa}{datos_sucios}{reset}")

# Limpiar producto
datos_sucios['producto'] = datos_sucios['producto'].str.strip().str.lower()

print(f"\n{azul}Producto limpio (minúsculas y sin espacios){reset}")
print(f"{turquesa}{datos_sucios['producto']}{reset}")

# Limpiar precio
datos_sucios['precio'] = (
    datos_sucios['precio']
    .str.replace('€', '', regex=False)
    .astype(float)
)

print(f"\n{azul}Precio convertido a float{reset}")
print(f"{turquesa}{datos_sucios['precio']}{reset}")

# Limpiar stock
datos_sucios['stock'] = (
    datos_sucios['stock']
    .replace('agotado', '0')
    .astype(int)
)

print(f"\n{azul}Stock limpio{reset}")
print(f"{turquesa}{datos_sucios['stock']}{reset}")

# Estandarizar categoría
datos_sucios['categoria'] = (
    datos_sucios['categoria']
    .str.lower()
    .replace({'informatica': 'Informática', 'pantallas': 'Pantallas'})
)

print(f"\n{azul}Categoría estandarizada{reset}")
print(f"{turquesa}{datos_sucios['categoria']}{reset}")

# Eliminar duplicados manteniendo el menor precio
datos_limpios = datos_sucios.sort_values('precio').drop_duplicates(
    subset='producto', keep='first'
)

print(f"\n{magenta}DataFrame final limpio (sin duplicados){reset}")
print(f"{lima}{datos_limpios}{reset}")

````

## 🔴 Ejercicio 10: Análisis Completo de Ventas (Caso Real)

Analiza un dataset de ventas de una tienda online:

```python
np.random.seed(42)
n = 100

ventas_completo = pd.DataFrame({
    'Fecha': pd.date_range('2024-01-01', periods=n, freq='D'),
    'Producto': np.random.choice(['Laptop', 'Mouse', 'Teclado', 'Monitor', 'Auriculares'], n),
    'Cantidad': np.random.randint(1, 10, n),
    'Precio_Unitario': np.random.choice([599.99, 19.99, 45.50, 249.99, 79.99], n),
    'Descuento': np.random.choice([0, 0.05, 0.10, 0.15], n),
    'Cliente_Tipo': np.random.choice(['Nuevo', 'Recurrente'], n, p=[0.3, 0.7])
})
```

Realiza un análisis completo:
1. Crea una columna 'Ingreso_Total' = Cantidad × Precio_Unitario × (1 - Descuento)
2. Calcula el ingreso total por tipo de producto
3. Encuentra el producto más vendido (por cantidad de unidades)
4. Calcula el ticket promedio por tipo de cliente
5. Identifica el mes con mayores ingresos
6. Calcula qué porcentaje de las ventas tuvo descuento
7. Crea un resumen ejecutivo con las métricas más importantes

````python
# Título: Análisis Completo de Ventas (Caso Real)
# Autor: Ivana Sánchez Pérez

# colores
rojo = '\033[91m'
verde = '\033[92m'
azul = '\033[94m'
magenta = '\033[95m'
amarillo = '\033[93m'
rosa = '\033[38;5;200m'
turquesa = '\033[38;5;44m'
azul_marino = '\33[38;5;67m'
lima = '\33[38;5;46m'
reset = '\033[0m'

import pandas as pd
import numpy as np

# Crear el DataFrame
np.random.seed(42)
n = 100

ventas_completo = pd.DataFrame({
    'Fecha': pd.date_range('2024-01-01', periods=n, freq='D'),
    'Producto': np.random.choice(['Laptop', 'Mouse', 'Teclado', 'Monitor', 'Auriculares'], n),
    'Cantidad': np.random.randint(1, 10, n),
    'Precio_Unitario': np.random.choice([599.99, 19.99, 45.50, 249.99, 79.99], n),
    'Descuento': np.random.choice([0, 0.05, 0.10, 0.15], n),
    'Cliente_Tipo': np.random.choice(['Nuevo', 'Recurrente'], n, p=[0.3, 0.7])
})

print(f"\n{azul}DataFrame de ventas{reset}")
print(f"{turquesa}{ventas_completo.head()}{reset}")

# Crear columna Ingreso_Total
ventas_completo['Ingreso_Total'] = (
    ventas_completo['Cantidad'] *
    ventas_completo['Precio_Unitario'] *
    (1 - ventas_completo['Descuento'])
)

print(f"\n{azul}Ingreso total por venta{reset}")
print(f"{turquesa}{ventas_completo[['Cantidad','Precio_Unitario','Descuento','Ingreso_Total']].head()}{reset}")

# Ingreso total por tipo de producto
ingreso_producto = ventas_completo.groupby('Producto')['Ingreso_Total'].sum()

print(f"\n{azul}Ingreso total por producto{reset}")
print(f"{lima}{ingreso_producto}{reset}")

# Producto más vendido (por cantidad)
producto_mas_vendido = ventas_completo.groupby('Producto')['Cantidad'].sum().idxmax()

print(f"\n{verde}Producto más vendido (unidades){producto_mas_vendido}{reset}")

# Ticket promedio por tipo de cliente
ticket_promedio = ventas_completo.groupby('Cliente_Tipo')['Ingreso_Total'].mean()

print(f"\n{azul}Ticket promedio por tipo de cliente{reset}")
print(f"{turquesa}{ticket_promedio}{reset}")

# Mes con mayores ingresos
ventas_completo['Mes'] = ventas_completo['Fecha'].dt.month
ingresos_mes = ventas_completo.groupby('Mes')['Ingreso_Total'].sum()
mes_mayor_ingreso = ingresos_mes.idxmax()

print(f"\n{magenta}Mes con mayores ingresos: {mes_mayor_ingreso}{reset}")

# Porcentaje de ventas con descuento
porcentaje_descuento = (ventas_completo['Descuento'] > 0).mean() * 100

print(f"\n{amarillo}Porcentaje de ventas con descuento: {porcentaje_descuento:.2f}%{reset}")

# Resumen ejecutivo
print(f"\n{rojo}RESUMEN EJECUTIVO{reset}")
print(f"{lima}Ingresos totales: {ventas_completo['Ingreso_Total'].sum():.2f} €{reset}")
print(f"{lima}Producto más vendido: {producto_mas_vendido}{reset}")
print(f"{lima}Mes con más ingresos: {mes_mayor_ingreso}{reset}")
print(f"{lima}Ventas con descuento: {porcentaje_descuento:.2f}%{reset}")
````
