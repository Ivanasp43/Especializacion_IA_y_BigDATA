## Demo básica sin predicción 



```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, LSTM

# Preparación de los datos - 3 pasos de tiempo, 1 característica
datos = np.array([0.5, 0.4, 0.3]).reshape((1, 3, 1))

# Creamos la arquitectura base
entrada = Input(shape=(3,1))  # Define que miramos 3 pasos al pasado con 1 dato cada vez (temperatura de las 3 últimas horas)  
lstm_out = LSTM(5)(entrada) # 5 unidades (neuronas) de memoria
modelo = Model(inputs=entrada, outputs=lstm_out) # La Api funcional 

print('Información del modelo:')
print()
print('-'*90)
print(modelo.summary()) # Muestra los parámetros internos que genera la red por las puertas lógicas
print('-'*90)
print('Tamaño de entrada: ', modelo.input_shape)
print('Tamaño de salida ', modelo.output_shape)
```

## Demo básica con predicción

## Demo básica con return_sequences

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, LSTM
import os
os.environ['TF_CPP_MIN_LOG_LEVEL'] = '2'

# Preparación de los datos - 3 pasos de tiempo, 1 característica
datos = np.array([0.5, 0.4, 0.3]).reshape((3, 1))

# Creamos la arquitectura base
entrada = Input(shape=(3,1))    
lstm_out = LSTM(5,
                return_sequences=True)(entrada) # en lugar del resumen final nos da una salida de cada paso de tiempo
modelo = Model(inputs=entrada, outputs=lstm_out)

prediccion = modelo.predict(datos, verbose=0)

print('Información del modelo:')
print('-'*90)
print(modelo.summary())

print('-'*105)
print('Tamaño de entrada: ', modelo.input_shape)
print('Tamaño de salida ', modelo.output_shape)

print()
print('-'*105)
print('Predicción (h_t): ', prediccion)
print('Tamaño predicción: ', prediccion.shape)
```

## Demo básica con return_state

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, LSTM
import os
os.environ['TF_CPP_MIN_LOG_LEVEL'] = '2'

# Preparación de los datos - 3 pasos de tiempo, 1 característica
datos = np.array([0.5, 0.4, 0.3]).reshape((3, 1))

# Creamos la arquitectura base
entrada = Input(shape=(3,1))    
lstm_out, h_t, c_t = LSTM(5,
                return_state=True)(entrada) # "Abrimos" la celda para ver tanto la memoria a corto plazo (H) como la de largo plazo (C)
modelo = Model(inputs=entrada, outputs=[lstm_out, h_t, c_t])

prediccion, h, c = modelo.predict(datos, verbose=0)

print('Información del modelo:')
print('-'*90)
print(modelo.summary())

print('-'*105)
print('Tamaño de entrada: ', modelo.input_shape)
print('Tamaño de salida ', modelo.output_shape)

print()
print('-'*105)
print('Predicción (lstm_out): ', prediccion)
print('Tamaño predicción "lstm_out": ', prediccion.shape)

print()
print('-'*105)
print('Predicción (h_t): ', h)
print('Tamaño predicción: ', prediccion.shape)

print()
print('-'*105)
print('Predicción (c_t): ', c)
print('Tamaño predicción "c_t": ', prediccion.shape)
```

## Demo básica: combinación return_sequences-return_state


```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, LSTM
import os
os.environ['TF_CPP_MIN_LOG_LEVEL'] = '2'

# Preparación de los datos - 3 pasos de tiempo, 1 característica
datos = np.array([0.5, 0.4, 0.3]).reshape((3, 1))

# Creamos la arquitectura base
entrada = Input(shape=(3,1))    
lstm_out, h, c = LSTM(5,
                        return_sequences=True,
                        return_state=True)(entrada)
modelo = Model(inputs=entrada, outputs=[lstm_out, h, c])

prediccion, h, c = modelo.predict(datos, verbose=0)

print('Información del modelo:')
print('-'*90)
print(modelo.summary())

print('-'*105)
print('Tamaño de entrada: ', modelo.input_shape)
print('Tamaño de salida ', modelo.output_shape)

print()
print('-'*105)
print('Predicción (lstm_out): ', prediccion)
print('Tamaño predicción "lstm_out": ', prediccion.shape)

print()
print('-'*105)
print('Predicción (h_t): ', h)
print('Tamaño predicción: ', prediccion.shape)

print()
print('-'*105)
print('Predicción (c_t): ', c)
print('Tamaño predicción "c_t": ', prediccion.shape)


```

## Demo básica LSTM + CAPA DENSE. Con función Sigmoidal

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, LSTM, Dense

# Preparación de los datos - 3 pasos de tiempo, 1 característica
datos = np.array([0.5, 0.4, 0.3]).reshape((3, 1))

# Creamos la arquitectura base
entrada = Input(shape=(3,1))    
lstm_out = LSTM(5)(entrada) 
modelo = Model(inputs=entrada, outputs=lstm_out)

prediccion = modelo.predict(datos, verbose=0)

# Agregar capa de salida Dense
salida = Dense(1, activation='sigmoid')(lstm_out) # Traduce la memoria a una probabilidad entre 0 y 1

print()
print('Información del modelo:')
print()
print('-'*90)
print(modelo.summary())

print()
print('-'*105)
print('Tamaño de entrada: ', modelo.input_shape)
print('Tamaño de salida ', modelo.output_shape)

print()
print('-'*105)
print('Predicción (clasificación binaria): ', prediccion)
print('Tamaño predicción: ', prediccion.shape)
```

## Demo básica LSTM + CAPA DENSE. Con función Softmax

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, LSTM, Dense

# Preparación de los datos - 3 pasos de tiempo, 1 característica
datos = np.array([0.5, 0.4, 0.3]).reshape((3, 1))

# Creamos la arquitectura base
entrada = Input(shape=(3,1))    
lstm_out = LSTM(5)(entrada) 
modelo = Model(inputs=entrada, outputs=lstm_out)

prediccion = modelo.predict(datos, verbose=0)

# Agregar capa de salida Dense
salida = Dense(1, activation='softmax')(lstm_out)  # Regresión multiclase: elige la categoriía más probable entre varias opciones

print()
print('Información del modelo:')
print()
print('-'*90)
print(modelo.summary())

print()
print('-'*105)
print('Tamaño de entrada: ', modelo.input_shape)
print('Tamaño de salida ', modelo.output_shape)

print()
print('-'*105)
print('Predicción (clasificación multiclase): ', prediccion)
print('Tamaño predicción: ', prediccion.shape)
```
