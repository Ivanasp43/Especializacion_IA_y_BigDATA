## 🟢 Ejercicio 1: Escritor y Lector de Notas

Crea dos funciones:

1. `guardar_nota(archivo, texto)`: Añade una línea al archivo con timestamp
2. `leer_notas(archivo)`: Lee y muestra todas las notas

Formato de cada nota:
```
[2025-01-15 10:30] Mi primera nota
[2025-01-15 10:31] Mi segunda nota
```

**Pista:** Usa `datetime.now().strftime("%Y-%m-%d %H:%M")`

````python
# Título: Escritor y lector de notas
# Autor: Ivana Sánchez Pérez

# colores
rojo = '\033[91m'
verde = '\033[92m'
azul = '\033[94m'
magenta = '\033[95m'
amarillo = '\033[93m'
rosa = '\033[38;5;200m'
turquesa = '\033[38;5;44m'
azul_marino = '\33[38;5;67m'
lima = '\33[38;5;46m'
reset = '\033[0m'

print(f"\n{rosa}   ===== Listado de notas ====={reset}")
print()

from datetime import datetime
formato = "%Y-%m-%d %H:%M"

def guardar_nota(archivo, texto):
    timestamp = datetime.now().strftime(formato)
    linea_nota = f"[{timestamp}] {texto}\n"
    with open(archivo, 'a', encoding='utf-8') as f:
        f.write(linea_nota)
    
def leer_notas(archivo):
    with open(archivo, 'r', encoding='utf-8') as f:
        contenido = f.read()
        print(f"{turquesa}{contenido.strip()}{reset}")
        
nombre_archivo = "notas.txt"

guardar_nota(nombre_archivo, "Mi primera nota")
guardar_nota(nombre_archivo, "Mi segunda nota")
leer_notas(nombre_archivo)
````

Crea una función `contar_palabras(archivo)` que:

## 🟡 Ejercicio 3: Gestor de Lista de Tareas (TODO)

Crea un sistema de tareas guardado en `tareas.txt`:

Funciones necesarias:
- `agregar_tarea(tarea)`: Añade una tarea
- `listar_tareas()`: Muestra todas las tareas numeradas
- `completar_tarea(numero)`: Marca una tarea como completada (añade ✓)
- `eliminar_tarea(numero)`: Elimina una tarea

Formato del archivo:
```
[ ] Estudiar Python
[✓] Hacer ejercicios
[ ] Practicar con archivos
```

````python
# Título: Gestor de lista de tareas
# Autor: Ivana Sánchez Pérez

# colores
rojo = '\033[91m'
verde = '\033[92m'
azul = '\033[94m'
magenta = '\033[95m'
amarillo = '\033[93m'
rosa = '\033[38;5;200m'
turquesa = '\033[38;5;44m'
azul_marino = '\33[38;5;67m'
lima = '\33[38;5;46m'
reset = '\033[0m'

print(f"\n{rosa}===== Gestor de listas ====={reset}")
print()

nombre_archivo = "tareas.txt"

def obtener_lista():
    try:
        with open(nombre_archivo, 'r', encoding='utf-8') as f:
            tareas = [linea.strip() for linea in f.readlines()]
            return [t for t in tareas if t] 
    except FileNotFoundError:
        return []
    
def agregar_tarea(tarea):
    nueva_tarea = f"[ ] {tarea}\n" 
    
    with open(nombre_archivo, 'a', encoding='utf-8') as f:
        f.write(nueva_tarea)
    
    print(f"{turquesa}Tarea añadida:{reset} {tarea}")

def listar_tareas():
    tareas = obtener_lista()
    
    if not tareas:
        print(f"{rojo}¡No hay tareas!{reset}")
        return
    for i, tarea in enumerate(tareas):
        print(f"{amarillo}{i + 1}. {tarea}{reset}")
        
def completar_tarea(numero):    
    
    tareas = obtener_lista()
    indice = numero - 1
        
    if 0 <= indice < len(tareas):
        tarea_original = tareas[indice]
        tarea_completada = tarea_original.replace("[ ]", "[✓]", 1)
        tareas[indice] = tarea_completada
            
        with open(nombre_archivo, 'w', encoding='utf-8') as f:
            for tarea in tareas:
                f.write(tarea + '\n')
            
        print(f"{lima}{tarea_completada}{reset}")
        print(f"{lima}La tarea número {numero} ha sido completada.{reset}")
    else:
        print(f"{rojo}¡¡ERROR!! La tarea {numero} no existe.{reset}")

def eliminar_tarea(numero):

    tareas = obtener_lista()
    indice = numero - 1
    if 0 <= indice < len(tareas):
        tarea_eliminada = tareas.pop(indice)
        
        texto_eliminado = tarea_eliminada[4:]   
        with open(nombre_archivo, 'w', encoding='utf-8') as f:
            for tarea in tareas:
                f.write(tarea + '\n')
            
        print(f"{verde}La tarea {numero}, '{texto_eliminado}', ha sido eliminada {reset}")
    else:
        print(f"{rojo}¡¡ERROR!! La tarea {numero} no existe.{reset}")

with open(nombre_archivo, 'w') as f:
    f.write('')
                
agregar_tarea("Estudiar Python")
agregar_tarea("Hacer ejercicios")
agregar_tarea("Practicar con archivos")
print()
print(f"{rosa}---LISTADO DE TAREAS---{reset}")
listar_tareas() 
print()
print(f"{rosa}---FINALIZACIÓN DE TAREAS---{reset}")
completar_tarea(2)
print()
print(f"{rosa}---ELIMINACIÓN DE TAREAS---{reset}")
eliminar_tarea(1) 
print()
print(f"{rosa}---RESULTADO FINAL---{reset}")
listar_tareas()  

    
````

## 🟡 Ejercicio 4: Manejo Robusto de División

Crea una función `division_segura(a, b)` que:
- Intente dividir a entre b
- Maneje `ZeroDivisionError` si b es 0
- Maneje `TypeError` si los valores no son numéricos
- Maneje `ValueError` si hay problemas de conversión
- Siempre muestre un mensaje de "Operación finalizada"

Prueba con:
- `division_segura(10, 2)` → 5.0
- `division_segura(10, 0)` → Error controlado
- `division_segura("10", "abc")` → Error controlado

````python
# Título: manejo robuto de división
# Autor: Ivana Sánchez Pérez

# colores
rojo = '\033[91m'
verde = '\033[92m'
azul = '\033[94m'
magenta = '\033[95m'
amarillo = '\033[93m'
rosa = '\033[38;5;200m'
turquesa = '\033[38;5;44m'
azul_marino = '\33[38;5;67m'
lima = '\33[38;5;46m'
reset = '\033[0m'

print(f"\n{rosa}===== Manejo de divisiones ====={reset}")
print()

def division_segura(a, b):
    try:
        resultado = a / b
        print(f"{magenta}El resultado de {a}/{b} es: {reset}{resultado}")
    except ZeroDivisionError:
        print(f"{rojo}¡¡ERROR!! No se puede dividir entre cero.{reset}")
    except TypeError:
        print(f"{rojo}¡¡ERROR!! Los valores deben ser numéricos.{reset}")
    except ValueError:
        print(f"{rojo}¡¡ERROR!! Valor no válido para la operación.{reset}")
    finally:
        print(f"{lima}Operación finalizada.\n{reset}")

print(f"{azul}Caso 1: division_segura (10, 2){reset}")
division_segura(10, 2)       

print(f"{azul}Caso 2: division_segura (10, 0){reset}")
division_segura(10, 0)       

print(f"{azul}Caso 3: division_segura ('10', 'abc'){reset}")
division_segura("10", "abc") 
````

## 🟡 Ejercicio 5: Conversor de CSV a Diccionario

Crea un archivo CSV con datos de estudiantes:

```csv
nombre,edad,nota
Ana,22,8.5
Carlos,23,7.0
Elena,21,9.2
```

Función `leer_csv(archivo)` que retorne:
```python
[
    {"nombre": "Ana", "edad": 22, "nota": 8.5},
    {"nombre": "Carlos", "edad": 23, "nota": 7.0},
    {"nombre": "Elena", "edad": 21, "nota": 9.2}
]
```

**No uses el módulo csv**, hazlo manualmente con split().

````python
# Título: Conversor de CSV a diccionario
# Autor: Ivana Sánchez Pérez

# colores
rojo = '\033[91m'
verde = '\033[92m'
azul = '\033[94m'
magenta = '\033[95m'
amarillo = '\033[93m'
rosa = '\033[38;5;200m'
turquesa = '\033[38;5;44m'
azul_marino = '\33[38;5;67m'
lima = '\33[38;5;46m'
reset = '\033[0m'

print(f"\n{rosa}  ===== Conversor de CSV a diccionario ====={reset}")
print()

def leer_csv(archivo):
    lista_estudiantes = []
    
    with open(archivo, "r", encoding="utf-8") as f:
        lineas = f.read().strip().split('\n')
        
        if not lineas or len(lineas) < 2:
            return lista_estudiantes

    for linea_datos in lineas[1:]:
        valores = linea_datos.split(",")
        
        estudiante = {
            "nombre": valores[0],
            "edad": int(valores[1]),
            "nota": float(valores[2])
        }

        lista_estudiantes.append(estudiante)
        
    return lista_estudiantes

resultado = leer_csv("estudiantes.csv")

print(f"{lima}¡Conversión Exitosa!{reset}")
print()
for estudiante in resultado:
    print(f"{verde}  {estudiante}{reset}")
````

## 🟡 Ejercicio 6: Validador de Entrada con Reintentos

Crea una función `pedir_numero(mensaje, intentos=3)` que:
- Pida un número al usuario
- Permita X intentos
- Maneje `ValueError` si la entrada no es numérica
- Retorne el número o `None` si se agotan los intentos

Para simular input(), usa una lista de prueba:
```python
entradas_prueba = ["abc", "12.5", "xyz"]
```

````python
# Título: Validador de entrada con reintentos
# Autor: Ivana Sánchez Pérez

# colores
rojo = '\033[91m'
verde = '\033[92m'
azul = '\033[94m'
magenta = '\033[95m'
amarillo = '\033[93m'
rosa = '\033[38;5;200m'
turquesa = '\033[38;5;44m'
azul_marino = '\33[38;5;67m'
lima = '\33[38;5;46m'
reset = '\033[0m'

print(f"\n{rosa}===== Validador de entradas ====={reset}")
print()

def pedir_numero(mensaje, entradas_prueba, intentos=3):
    
    for i in range(intentos):
        
        if i >= len(entradas_prueba):
            print(f"{rojo}¡¡ERROR!! Se agotaron las entradas de prueba.{reset}")
            break 
            
        entrada = entradas_prueba[i]
        
        try:
            print(f"{mensaje} (Intento {i + 1}/{intentos}): {entrada}")
            
            numero = float(entrada)
            return numero 
            
        except ValueError:
            print(f"{rojo}¡¡ERROR!! '{entrada}' no es un número válido.{reset}")
            
    print(f"{amarillo}¡Se agotaron todos los intentos!{reset}")
    return None


# pruebas
prueba1_entradas = ["abc", "12.5", "xyz"]
prueba2_entradas = ["abc", "def", "ghi"]
prueba3_entradas = ["15", "abc", "xyz"]

print(f"\n{rosa}##### Prueba 1 #####{reset}")
print(f"{rosa}--------------------{reset}")
resultado1 = pedir_numero(f"{azul}Por favor ingresa un número{reset}", prueba1_entradas)
print(f"{lima}Resultado: {resultado1}{reset}")

print(f"\n{rosa}##### Prueba 2 #####{reset}")
print(f"{rosa}--------------------{reset}")
resultado2 = pedir_numero(f"{azul}Ingresa un número{reset}", prueba2_entradas, intentos=2)
print(f"{lima}Resultado: {resultado2}{reset}")

print(f"\n{rosa}##### Prueba 3 #####{reset}")
print(f"{rosa}--------------------{reset}")
resultado3 = pedir_numero(f"{azul}Dame un número{reset}", prueba3_entradas)
print(f"{lima}Resultado: {resultado3}{reset}")
````

## 🔴 Ejercicio 7: Sistema de Logs con Loguru

Crea un sistema de logging usando la librería `loguru`:

**Instalación previa:**
```bash
uv add loguru
```

Funciones necesarias:
- `log(mensaje, nivel="INFO")`: Escribe en `app.log` usando loguru
  - Formato: `[2025-01-15 10:30:25] [INFO] Mensaje aquí`
  - Niveles: INFO, WARNING, ERROR
  - Añade automáticamente al final del archivo

- `leer_logs(nivel=None)`: Lee el archivo de logs
  - Si se especifica nivel, filtre solo esos logs
  - Retorne una lista de logs

**Ventajas de loguru:**
- Manejo automático de archivos (no necesitas abrir/cerrar)
- Formato elegante y configurable
- Rotación de logs automática (opcional)
- Mucho más profesional que logging manual

**Ejemplo de uso:**
```python
from loguru import logger

logger.info("Aplicación iniciada")
logger.warning("Memoria baja")
logger.error("Fallo en conexión")
```

````python
# Título: Sistema de logs con loguru
# Autor: Ivana Sánchez Pérez

# colores
rojo = '\033[91m'
verde = '\033[92m'
azul = '\033[94m'
magenta = '\033[95m'
amarillo = '\033[93m'
rosa = '\033[38;5;200m'
turquesa = '\033[38;5;44m'
azul_marino = '\33[38;5;67m'
lima = '\33[38;5;46m'
reset = '\033[0m'

print(f"\n{rosa}          ======= Sistema Loguru ======={reset}")
print()

from loguru import logger
logger.remove()

logger.add(
    "app.log",
    format="[{time:YYYY-MM-DD HH:mm:ss}] [{level}] {message}",
    level="INFO",
    rotation="10 MB" 
)

def log(mensaje: str, nivel: str = "INFO"):
   
    nivel = nivel.upper()
    if nivel == "INFO":
        logger.info(mensaje)
    elif nivel == "WARNING":
        logger.warning(mensaje)
    elif nivel == "ERROR":
        logger.error(mensaje)
    else:
        logger.info(f"[NIVEL_NO_VALIDO - {nivel}] {mensaje}")

def leer_logs(nivel: str = None) -> list[str]:
    
    log_file = "app.log"
    logs_filtrados = []

    with open(log_file, "r", encoding="utf-8") as f:
        lineas = f.readlines()
            
        if nivel:
            nivel_a_buscar = f"[{nivel.upper()}]"
                
            for linea in lineas:
                if nivel_a_buscar in linea:
                    logs_filtrados.append(linea.strip())
        else:
            logs_filtrados = [linea.strip() for linea in lineas]
        
    return logs_filtrados

# Ejemplo de uso

print(f"{azul_marino}Generando logs de prueba...{reset}")
log("Aplicación iniciada", "INFO")
log("Memoria baja", "WARNING")
log("Fallo en conexión", "ERROR")
log("Tarea de limpieza completada") 

print(f"\n{azul_marino}Leyendo TODOS los logs:{reset}")
todos_los_logs = leer_logs()
for linea in todos_los_logs:
    print(linea)

print(f"\n{azul_marino}Leyendo logs solo de nivel ERROR:{reset}")
logs_error = leer_logs(nivel="ERROR")
for linea in logs_error:
    print(linea)

print(f"\n{azul_marino}Leyendo logs solo de nivel WARNING:{reset}")
logs_warning = leer_logs(nivel="WARNING")
for linea in logs_warning:
    print(linea)
````

## 🔴 Ejercicio 8: Copia de Seguridad de Archivos

Crea una función `backup(archivo_origen)` que:
- Lea el archivo origen
- Cree una copia con el formato: `archivo_backup_YYYYMMDD_HHMMSS.ext`
- Maneje todas las excepciones posibles:
  - `FileNotFoundError`: archivo no existe
  - `PermissionError`: sin permisos
  - `IOError`: error de E/S
- Retorne `True` si tuvo éxito, `False` en caso contrario

Ejemplo:
- `notas.txt` → `notas_backup_20250115_103025.txt`

````python
# Título: Copia de seguridad de archivos
# Autor: Ivana Sánchez Pérez

# colores
rojo = '\033[91m'
verde = '\033[92m'
azul = '\033[94m'
magenta = '\033[95m'
amarillo = '\033[93m'
rosa = '\033[38;5;200m'
turquesa = '\033[38;5;44m'
azul_marino = '\33[38;5;67m'
lima = '\33[38;5;46m'
reset = '\033[0m'

print(f"\n{rosa}            ======= Copia de Seguridad ======={reset}")
print()

from datetime import datetime
import os

def backup(archivo_origen):
    
    try:
        nombre, extension = os.path.splitext(archivo_origen)
        fecha_hora = datetime.now().strftime("%Y%m%d_%H%M%S")
        archivo_backup = f"{nombre}_backup_{fecha_hora}{extension}"
        
        with open(archivo_origen, "rb") as original:
            contenido = original.read()
    
        with open(archivo_backup, "wb") as copia:
            copia.write(contenido)
        
        print(f"{verde}Copia de seguridad creada como '{archivo_backup}'{reset}")
        return True

    except FileNotFoundError:
        print(f"{rojo}¡¡ERROR!! El archivo no existe.{reset} (FileNotFoundError)")   
    except PermissionError:
        print(f"{rojo}¡¡ERROR!! No tienes permisos para acceder o crear el archivo.{reset} (PermissionError)")   
    except IOError:
        print(f"{rojo}¡¡ERROR!! Error de entrada/salida al manejar el archivo.{reset} (IOError)")
    
    return False

nombre_prueba = "notas.txt"
try:
    with open(nombre_prueba, "w") as f:
        f.write("Contenido para el backup.")
except:
    pass

print(f"\n{azul}--- Prueba sin error ({nombre_prueba}) ---{reset}")
resultado_exito = backup(nombre_prueba)
print(f"{azul_marino}Resultado Final:{reset} {resultado_exito}")

print(f"\n{azul}--- Prueba con error (Archivo no existe) ---{reset}")
resultado_error = backup("archivo_inexistente.xyz")
print(f"{azul_marino}Resultado Final:{reset} {resultado_error}")
````

````python
# Título: Analizador de archivo JSON
# Autor: Ivana Sánchez Pérez

# colores
rojo = '\033[91m'
verde = '\033[92m'
azul = '\033[94m'
magenta = '\033[95m'
amarillo = '\033[93m'
rosa = '\033[38;5;200m'
turquesa = '\033[38;5;44m'
azul_marino = '\33[38;5;67m'
lima = '\33[38;5;46m'
reset = '\033[0m'

print(f"\n{rosa}======= Análisis del JSON ======={reset}")
print()

import json
import os

nombre_json = "datos.json"

def cargar_json(archivo):
    with open(archivo, "r", encoding="utf-8") as f:
        return json.load(f)

def guardar_json(archivo, datos):

    with open(archivo, "w", encoding="utf-8") as f:
        json.dump(datos, f, indent=4, ensure_ascii=False)
    return True

def crear_archivo_json(nombre_archivo):
    datos_iniciales = {
        "usuarios": [
            {"nombre": "Ana", "edad": 25, "ciudad": "Madrid"},
            {"nombre": "Carlos", "edad": 30, "ciudad": "Barcelona"},
            {"nombre": "Elena", "edad": 28, "ciudad": "Madrid"}
        ]
    }
    print(f"{azul}***Inicializando '{nombre_archivo}'...{reset}")
    guardar_json(nombre_archivo, datos_iniciales)


def agregar_usuario(archivo, nuevo_usuario):
    
    datos = cargar_json(archivo)
    datos["usuarios"].append(nuevo_usuario)
    
    if guardar_json(archivo, datos):
        print(f"{azul}***Guardando datos en '{archivo}'{reset}")
        print(f"\n{turquesa}Usuario {nuevo_usuario['nombre']} agregado.{reset}")
        return True
    return False

def buscar_por_ciudad(archivo, ciudad):
    
    # Imprimimos el mensaje de carga aquí
    print(f"\n{azul}***Cargando el archivo'{archivo}' para buscar...{reset}")
    datos = cargar_json(archivo)
    usuarios_en_ciudad = []
    
    for usuario in datos["usuarios"]:
        if usuario.get("ciudad") == ciudad:
            usuarios_en_ciudad.append(usuario)
            
    return usuarios_en_ciudad

# DEMOSTRACIONES 

# Limpieza inicial
if os.path.exists(nombre_json):
    os.remove(nombre_json)

#Inicializar el archivo
crear_archivo_json(nombre_json)

# Agregar nuevo usuario
nuevo_usuario = {"nombre": "David", "edad": 22, "ciudad": "Barcelona"}
agregar_usuario(nombre_json, nuevo_usuario)

# Buscar usuarios por ciudad
ciudad_a_buscar = "Madrid"
usuarios_madrid = buscar_por_ciudad(nombre_json, ciudad_a_buscar) 

print(f"\n{turquesa}Usuarios encontrados en {ciudad_a_buscar}:{reset}")
print(f"{turquesa}-------------------------------{reset}")
for u in usuarios_madrid:
    print(f"{lima}  - {u['nombre']} ({u['edad']} años){reset}")

#Contenido total
print(f"\n{azul}***Cargando el archivo '{nombre_json}' para conteo final...{reset}")
datos_finales = cargar_json(nombre_json)

print()
print(f"{turquesa}Total de usuarios finales: {len(datos_finales['usuarios'])}{reset}")

# Limpieza
if os.path.exists(nombre_json):
    os.remove(nombre_json)
````

## 🎯 Ejercicio Bonus: Gestor de Configuración

Crea un sistema para gestionar archivos de configuración:

Archivo `config.ini`:
```ini
[DATABASE]
host = localhost
port = 5432
user = admin

[APP]
debug = True
max_users = 100
```

Funciones:
- `leer_config(archivo)`: Retorna dict con la configuración
- `obtener_valor(seccion, clave)`: Obtiene un valor específico
- `actualizar_valor(seccion, clave, valor)`: Actualiza un valor

**Bonus:** Convierte automáticamente tipos (True → bool, 100 → int)

````python
# Título: Gestor de configuración
# Autor: Ivana Sánchez Pérez

# colores
rojo = '\033[91m'
verde = '\033[92m'
azul = '\033[94m'
magenta = '\033[95m'
amarillo = '\033[93m'
rosa = '\033[38;5;200m'
turquesa = '\033[38;5;44m'
azul_marino = '\33[38;5;67m'
lima = '\33[38;5;46m'
reset = '\033[0m'

print(f"\n{rosa}======= Gestor de Configuración  ======={reset}")
print()

import os
import configparser

configuracion = "config.ini"

def convertir_tipo(valor_str):
    
    valor_str = valor_str.strip()
    if valor_str.lower() == 'true':
        return True
    if valor_str.lower() == 'false':
        return False
    
    # Conversión numérica   
    try:
        return int(valor_str)
    except ValueError:
        pass 
        
    #Conversión de punto flotante 
    try:
        return float(valor_str)
    except ValueError:
        pass 
        
    return valor_str

def crear_configuracion_inicial(archivo):
    config = configparser.ConfigParser()
    
    config['DATABASE'] = {
        'host': 'localhost',
        'port': '5432',
        'user': 'admin'
    }
    
    config['APP'] = {
        'debug': 'True', 
        'max_users': '100'
    }
    
    with open(archivo, 'w') as configfile:
        config.write(configfile)
    print(f"{turquesa} Archivo '{archivo}' creado.{reset}")

def leer_config(archivo):
    config = configparser.ConfigParser()
    config.read(archivo)
    
    # Converir a un diccionario normal
    config_dict = {}
    for seccion in config.sections():
        config_dict[seccion] = {}
        for clave, valor_str in config.items(seccion):
            config_dict[seccion][clave] = convertir_tipo(valor_str)
            
    return config_dict

def obtener_valor(seccion, clave, archivo=configuracion):
    config_data = leer_config(archivo)
    
    if seccion in config_data and clave in config_data[seccion]:
        return config_data[seccion][clave]
    
    return None

def actualizar_valor(seccion, clave, valor, archivo=configuracion):
    config = configparser.ConfigParser()
    config.read(archivo)
    
    if seccion not in config:
        config[seccion] = {}
        
    config[seccion][clave] = str(valor) 
    
    with open(archivo, 'w') as configfile:
        config.write(configfile)
        
    return True

if __name__ == "__main__":
    
    if os.path.exists(configuracion):
        os.remove(configuracion)
    crear_configuracion_inicial(configuracion)
    
    print(f"\n{azul}\n--- Lectura y Conversión de Tipos ---{reset}")
    
    configuracion_completa = leer_config(configuracion)
    print(f"\n{azul}Configuración (Diccionario Python):{reset}\njson.dumps(configuracion_completa, indent=2)")

    # Obtener valores y verificar tipos
    debug_val = obtener_valor('APP', 'debug')
    max_users_val = obtener_valor('APP', 'max_users')
    host_val = obtener_valor('DATABASE', 'host')
    
    print(f"\nValor 'debug': {debug_val}, Tipo: {type(debug_val)}")
    print(f"Valor 'max_users': {max_users_val}, Tipo: {type(max_users_val)}")
    print(f"Valor 'host': {host_val}, Tipo: {type(host_val)}")
    
    print(f"\n{azul}--- Actualización de Valores ---{reset}")
    
    actualizar_valor('APP', 'debug', False)
    actualizar_valor('DATABASE', 'port', 5433)
    actualizar_valor('CUSTOM', 'version', 2.0)

    print(f"Nuevo 'debug': {obtener_valor('APP', 'debug')} (Tipo: {type(obtener_valor('APP', 'debug'))})")
    print(f"Nuevo 'port': {obtener_valor('DATABASE', 'port')} (Tipo: {type(obtener_valor('DATABASE', 'port'))})")
    print(f"Nuevo 'version': {obtener_valor('CUSTOM', 'version')} (Tipo: {type(obtener_valor('CUSTOM', 'version'))})")

    if os.path.exists(configuracion):
        os.remove(configuracion)
````
