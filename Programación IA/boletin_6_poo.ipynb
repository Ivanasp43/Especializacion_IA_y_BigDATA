````python
# Título: POO. Clase estudiante
# Autor: Ivana Sánchez Pérez

# colores
rojo = '\033[91m'
verde = '\033[92m'
azul = '\033[94m'
magenta = '\033[95m'
amarillo = '\033[93m'
rosa = '\033[38;5;200m'
turquesa = '\033[38;5;44m'
azul_marino = '\33[38;5;67m'
lima = '\33[38;5;46m'
reset = '\033[0m'

print(f"\n{azul}===== Pruebas con objetos estudiante ====={reset}")
print()

class Estudiante:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad
        self.calificaciones = []
    def agregar_calificacion(self, nota):
        if 0 <= nota <= 10:
            self.calificaciones.append(nota)
            print(f"{azul_marino}Nota {nota} agregada a {self.nombre}.{reset}")
        else:
            print(f"{rojo}¡¡ERROR!! La nota debe estar entre 0 y 10.{reset}")
            
    def promedio(self):
        if not self.calificaciones:
            return 0
        return sum(self.calificaciones) / len(self.calificaciones)
    
    def mejor_nota(self):
        if not self.calificaciones:
            return 0
        return max(self.calificaciones)
    
    def esta_aprobado(self):
        return self.promedio() >=5
    
    def __str__(self):
        return (f"estudiantes: {self.nombre} | Edad: {self.edad} años | "
                f"Calificaciones: {self.calificaciones}")
        
# Crear un  estudiante
estudiante1 = Estudiante("Ana Pérez", 20)
# Agregar calificaciones
estudiante1.agregar_calificacion(8)
estudiante1.agregar_calificacion(6.5)
estudiante1.agregar_calificacion(4) 
estudiante1.agregar_calificacion(9)

# Probar todos los métodos
print(f"\n{rosa}Resultados de {estudiante1.nombre}{reset}")
print(f"{turquesa}Promedio de calificaciones:{reset}{amarillo} {estudiante1.promedio():.2f}{reset}")
print(f"{turquesa}Mejor nota:{reset}{amarillo} {estudiante1.mejor_nota()}{reset}")
print(f"{turquesa}¿Está aprobado?:{reset}{amarillo} {estudiante1.esta_aprobado()}")
print()

# Prueba con otro estudiante y caso de reprobación
estudiante2 = Estudiante("Juan Gómez", 22)
estudiante2.agregar_calificacion(3)
estudiante2.agregar_calificacion(4.5)
estudiante2.agregar_calificacion(5) 
estudiante2.agregar_calificacion(2)
print(f"\n{rosa}Resultados de {estudiante2.nombre}{reset}")
print(f"{turquesa}Promedio de calificaciones:{reset}{amarillo} {estudiante2.promedio():.2f}{reset}")
print(f"{turquesa}Mejor nota:{reset}{amarillo} {estudiante2.mejor_nota()}{reset}")
print(f"{turquesa}¿Está aprobado?:{reset}{amarillo} {estudiante2.esta_aprobado()}")
````

## 🟡 Ejercicio 3: Clase CuentaBancaria

Crea una clase `CuentaBancaria` con:

**Atributos:**
- `titular`
- `saldo` (privado, usa `_saldo`)
- `numero_cuenta` (generado automáticamente)

**Métodos:**
- `depositar(cantidad)`: añade al saldo
- `retirar(cantidad)`: resta del saldo (validar saldo suficiente)
- `transferir(otra_cuenta, cantidad)`: transfiere dinero
- `consultar_saldo()`: retorna el saldo
- `__str__()`: información de la cuenta

**Validaciones:**
- No permitir cantidades negativas
- No permitir retirar más del saldo disponible

````python
# Título: POO. Clase cuenta bancaria
# Autor: Ivana Sánchez Pérez

# colores
rojo = '\033[91m'
verde = '\033[92m'
azul = '\033[94m'
magenta = '\033[95m'
amarillo = '\033[93m'
rosa = '\033[38;5;200m'
turquesa = '\033[38;5;44m'
azul_marino = '\33[38;5;67m'
lima = '\33[38;5;46m'
reset = '\033[0m'

import random

print(f"\n{rosa}===== Pruebas objetos cuenta bancaria ====={reset}")
print()

class CuentaBancaria:
    def _generar_numero_cuenta():
        return str(random.randint(10000000, 99999999))
    def __init__(self, titular, saldo_inicial=0):
        self.titular = titular
        self._saldo = 0.0
        if saldo_inicial > 0:
            self._saldo = saldo_inicial
            
        self.numero_cuenta = CuentaBancaria._generar_numero_cuenta()
        print(f"{turquesa}Cuenta creada para {self.titular} (Nº: {self.numero_cuenta}){reset}")
        
    def consultar_saldo(self):
        return self._saldo
    
    def depositar(self, cantidad):
        if cantidad <= 0:
            print(f"{rojo}¡¡ERROR!! La cantidad debe ser positiva.{reset}")
            return False
        self._saldo += cantidad
        print(f"{lima}Depósito de ${cantidad:.2f} realizado. Saldo actual: ${self._saldo:.2f}{reset}")
        
    def retirar(self, cantidad):
        if cantidad <= 0:
            print(f"{rojo}¡¡ERROR!! La cantidad debe ser positiva.{reset}")
            return False
        
        if cantidad > self._saldo:
            print(f"{rojo}¡¡ERROR!! Saldo insuficiente. Disponible en cuenta: ${self._saldo:.2f}{reset}")
            return False
        
        self._saldo -= cantidad
        print(f"{lima}Retiro de ${cantidad:.2f} realizado. Saldo actual: ${self._saldo:.2f}{reset}")
        return True

    def transferir(self, otra_cuenta, cantidad):
        if not isinstance(otra_cuenta, CuentaBancaria):
            print(f"{rojo}¡¡ERROR!! El destino debe ser un objeto CuentaBancaria.{reset}")
            return False
            
        if cantidad <= 0:
            print(f"{rojo}¡¡ERROR!! La cantidad de la transferencia debe ser positiva.{reset}")
            return False

        if cantidad > self._saldo:
            print(f"{rojo}¡¡ERROR!! Saldo insuficiente para realizar la tranferencia ${cantidad:.2f}{reset}")
            return False

        self._saldo -= cantidad
        otra_cuenta.saldo += cantidad
        
        print(f"{lima}Transferencia de ${cantidad:.2f} a la cuenta {otra_cuenta.numero_cuenta} realizada.{reset}")
        print(f"{magenta}   Saldo actual de {self.titular}: ${self._saldo:.2f}{reset}")
        return True

    def __str__(self):
        return (f"{magenta}--- Información de Cuenta ---{reset}\n"
                f"{azul_marino}Titular: {self.titular}\n"
                f"N° Cuenta: {self.numero_cuenta}\n"
                f"Saldo: ${self._saldo:.2f}{reset}")

# Crear cuentas
cuenta_maria = CuentaBancaria("María López", saldo_inicial=1000.00)
cuenta_pedro = CuentaBancaria("Pedro García")

print("\n", cuenta_maria)
print("\n", cuenta_pedro)

# Prueba de Depósito y Validaciones
print(f"{magenta}\n--- Pruebas de Depósito ---{reset}")
cuenta_maria.depositar(500.50)
cuenta_maria.depositar(-100) 
print(f"Saldo de María después del depósito: ${cuenta_maria.consultar_saldo():.2f}")

# Prueba de Retiro y Validaciones
print(f"{magenta}\n--- Pruebas de Retiro ---{reset}")
cuenta_maria.retirar(200.00) 
cuenta_maria.retirar(1500.00) 
````

## 🟡 Ejercicio 4: Herencia - Sistema de Vehículos

Crea una jerarquía de clases:

**Clase base `Vehiculo`:**
- Atributos: `marca`, `modelo`, `año`
- Método: `describir()`

**Clases derivadas:**

**`Coche(Vehiculo)`:**
- Atributo adicional: `num_puertas`
- Sobrescribe `describir()` para incluir puertas

**`Moto(Vehiculo)`:**
- Atributo adicional: `tipo` (deportiva, scooter, etc.)
- Sobrescribe `describir()` para incluir tipo

Crea instancias de cada clase y muestra sus descripciones.

````python
# Título: POO. Herencia- Sistema de vehículos
# Autor: Ivana Sánchez Pérez

# colores
rojo = '\033[91m'
verde = '\033[92m'
azul = '\033[94m'
magenta = '\033[95m'
amarillo = '\033[93m'
rosa = '\033[38;5;200m'
turquesa = '\033[38;5;44m'
azul_marino = '\33[38;5;67m'
lima = '\33[38;5;46m'
reset = '\033[0m'

print(f"\n{azul}===== Prueba del sistema de vehículos ====={reset}")
print()

class Vehiculo:
    def __init__(self, marca, modelo, año):
        self.marca = marca
        self.modelo = modelo
        self.año = año
    def describir(self):
        return f"{turquesa}Vehículo: {self.marca} {self.modelo} del año {self.año}.{reset}"

class Coche(Vehiculo):
    def __init__(self, marca, modelo, año, num_puertas):
        super().__init__(marca, modelo, año)
        self.num_puertas = num_puertas   
    def describir(self):
        descripcion_base = super().describir()
        return f"{descripcion_base.strip(',')}. {amarillo}Es un coche con {self.num_puertas} puertas{reset}"

class Moto(Vehiculo):
    def __init__(self, marca, modelo, año, tipo):
        super().__init__(marca, modelo, año)
        self.tipo = tipo
    def describir(self):
        descripcion_base = super().describir()
        return f"{descripcion_base.strip('.')}. {amarillo}Es una moto de tipo {self.tipo}.{reset}"

# Instancia de la clase base
vehiculo_base = Vehiculo("Generico", "Modelo-X", 2020)
print(f"{rosa}Base:{reset} {amarillo}{vehiculo_base.describir()}{reset}")

# Instancia de la clase Coche
mi_coche = Coche("Renault", "Clio", 2023, 5)
print(f"{rosa}Coche:{reset} {amarillo}{mi_coche.describir()}{reset}")

# Instancia de la clase Moto
mi_moto = Moto("Honda", "CBR1000RR", 2024, "deportiva")
print(f"{rosa}Moto:{reset} {amarillo}{mi_moto.describir()}{reset}")

# Prueba de polimorfismo en una lista
print(f"\n{azul}--- Demostración de Polimorfismo ---{reset}")
garaje = [vehiculo_base, mi_coche, mi_moto]

for v in garaje:
    print(f"{rosa}* Tipo {type(v).__name__}:{reset}{amarillo} {v.describir()}{reset}")    
````

## 🟡 Ejercicio 5: Clase con Propiedades

Crea una clase `Temperatura` con:

**Atributo privado:**
- `_celsius`

**Propiedades (usa `@property`):**
- `celsius`: get/set para Celsius
- `fahrenheit`: get/set calculado automáticamente
- `kelvin`: get/set calculado automáticamente

Al cambiar cualquiera, las demás se actualizan automáticamente.

Ejemplo:
```python
temp = Temperatura()
temp.celsius = 25
print(temp.fahrenheit)  # 77.0
print(temp.kelvin)      # 298.15
```

````python
# Título: POO. Clases con propiedades
# Autor: Ivana Sánchez Pérez

# Definición de colores
rojo = '\033[91m'
verde = '\033[92m'
azul = '\033[94m'
magenta = '\033[95m'
amarillo = '\033[93m'
rosa = '\033[38;5;200m'
turquesa = '\033[38;5;44m'
azul_marino = '\33[38;5;67m'
lima = '\33[38;5;46m'
reset = '\033[0m'

print(f"\n{azul}=====  Clase Temperatura ====={reset}")
print()

class Temperatura:
    def __init__(self, celsius=0.0):
        self._celsius = float(celsius)
    @property
    def celsius(self):
        return self._celsius

    @celsius.setter
    def celsius(self, value):
        self._celsius = float(value)
        print(f"{rosa}Temperatura base establecida a {self._celsius}ºC.{reset}")
        
    # Propiedad fahrenheit (Calculado automáticamente)
    @property
    def fahrenheit(self):
        return (self._celsius * 1.8) + 32 # F=C*1.8+32

    @fahrenheit.setter # CORRECCIÓN 3: Eliminado el espacio extra
    def fahrenheit(self, value):
        self._celsius = (float(value) - 32) / 1.8 # C=(F-32)/1.8
        print (f"{rosa}Temperatura establecida a {float(value)}ºF. Base (ºC) actualizada.{reset}")

    # Propiedad kelvin (Calculado automáticamente)
    @property
    def kelvin(self):
        return self._celsius + 273.15  # K=C+273.15

    @kelvin.setter
    def kelvin(self, value):
        self._celsius = float(value) - 273.15
        print(f"{rosa}Temperatura establecida a {float(value)} K. Base (°C) actualizada.{reset}")


# Pruebas
print(f"{magenta}Cambiar Celsius y ver otros valores {reset}")
temp = Temperatura()
temp.celsius = 25 

print(f"{turquesa}Celsius:{reset}{amarillo} {temp.celsius}°C{reset}")      
print(f"{turquesa}Fahrenheit:{reset}{amarillo} {temp.fahrenheit}°F{reset}") 
print(f"{turquesa}Kelvin:{reset}{amarillo} {temp.kelvin:.2f} K{reset}")     

print(f"\n{magenta}Cambiar Fahrenheit y ver otros valores{reset}")
temp.fahrenheit = 98.6 

print(f"{turquesa}Celsius:{reset}{amarillo} {temp.celsius}°C{reset}")       
print(f"{turquesa}Fahrenheit:{reset}{amarillo} {temp.fahrenheit}°F{reset}")
print(f"{turquesa}Kelvin:{reset}{amarillo} {temp.kelvin:.2f} K{reset}") 

print(f"\n{magenta}Cambiar Kelvin y ver otros valores{reset}")
temp.kelvin = 0.0 

print(f"{turquesa}Celsius:{reset}{amarillo} {temp.celsius}°C{reset}")       
print(f"{turquesa}Fahrenheit:{reset}{amarillo} {temp.fahrenheit:.2f}°F{reset}") 
print(f"{turquesa}Kelvin:{reset}{amarillo} {temp.kelvin} K{reset}")        
````

## 🟡 Ejercicio 6: Métodos Estáticos y de Clase

Crea una clase `Matematicas` con:

**Atributo de clase:**
- `PI = 3.14159`
- `contador_operaciones = 0`

**Métodos estáticos (`@staticmethod`):**
- `es_par(n)`: retorna True si n es par
- `factorial(n)`: calcula factorial
- `maximo(a, b, c)`: retorna el máximo de tres números

**Métodos de clase (`@classmethod`):**
- `area_circulo(cls, radio)`: calcula área usando cls.PI
- `incrementar_contador(cls)`: incrementa contador_operaciones
- `obtener_contador(cls)`: retorna el contador

Prueba todos los métodos sin crear instancias.

````python
# Título: POO. Métodos estáticos y de clase
# Autor: Ivana Sánchez Pérez

# colores
rojo = '\033[91m'
verde = '\033[92m'
azul = '\033[94m'
magenta = '\033[95m'
amarillo = '\033[93m'
rosa = '\033[38;5;200m'
turquesa = '\033[38;5;44m'
azul_marino = '\33[38;5;67m'
lima = '\33[38;5;46m'
reset = '\033[0m'

class Matematicas:
    PI = 3.14159
    contador_operaciones = 0

    # Métodos estáticos 
    @staticmethod
    def es_par(n):
        return n % 2 == 0
    
    @staticmethod
    def factorial(n):
        if n < 0:
            return (f"{rojo}¡¡ERROR!! Factorial no definido para números negativos.{reset}")
        if n == 0:
            return 1
        
        resultado = 1
        for i in range(1, n + 1):
            resultado *= i  
        return resultado
    
    @staticmethod
    def maximo(a, b, c):
        return max(a, b, c)

    # Métodos de clase
    @classmethod
    def area_circulo(cls, radio):
        area = cls.PI * (radio ** 2) # Área = PI * radio^2
        cls.incrementar_contador() 
        return area

    @classmethod
    def incrementar_contador(cls):
        cls.contador_operaciones += 1
            
    @classmethod
    def obtener_contador(cls):
        return cls.contador_operaciones

# Pruebas

print(f"\n{azul}--- Prueba de Métodos Estáticos ---{reset}")
print(f"{turquesa}¿Es par el 10?{reset}{amarillo} {Matematicas.es_par(10)}{reset}") 
print(f"{turquesa}Factorial de 5:{reset}{amarillo} {Matematicas.factorial(5)}{reset}") 
print(f"{turquesa}Máximo de (10, 5, 20):{reset}{amarillo} {Matematicas.maximo(10, 5, 20)}{reset}") 

print(f"\n{azul}--- Prueba de Métodos de Clase y Atributos de Clase ---{reset}")
print(f"{turquesa}El valor de PI es:{reset}{amarillo} {Matematicas.PI}{reset}")
print(f"{turquesa}Contador de operaciones inicial:{reset}{amarillo} {Matematicas.obtener_contador()}{reset}") 

# Realizar una operación de clase
radio = 4
area = Matematicas.area_circulo(radio) 
print(f"\n{magenta}Área del círculo de radio {radio}:{reset}{amarillo} {area:.3f}{reset}") 

# Verificar el contador
print(f"{magenta}Contador de operaciones después del cálculo:{reset}{amarillo} {Matematicas.obtener_contador()}{reset}") 

# Realizar otra operación de clase
Matematicas.incrementar_contador()
print(f"{magenta}Contador después de incrementar directamente:{reset}{amarillo} {Matematicas.obtener_contador()}{reset}") 
````

## 🔴 Ejercicio 7: Sistema de Biblioteca

Crea un sistema completo con múltiples clases:

**Clase `Libro`:**
- Atributos: `titulo`, `autor`, `isbn`, `disponible`
- Métodos: `prestar()`, `devolver()`, `__str__()`

**Clase `Usuario`:**
- Atributos: `nombre`, `id`, `libros_prestados` (lista)
- Métodos: `tomar_prestado(libro)`, `devolver_libro(libro)`, `__str__()`

**Clase `Biblioteca`:**
- Atributos: `nombre`, `libros` (lista), `usuarios` (lista)
- Métodos:
  - `agregar_libro(libro)`
  - `registrar_usuario(usuario)`
  - `prestar_libro(isbn, id_usuario)`
  - `devolver_libro(isbn, id_usuario)`
  - `libros_disponibles()`
  - `buscar_libro(titulo)`

Crea una biblioteca, añade libros y usuarios, y simula préstamos.

````python
# Título: POO. Sistema de biblioteca
# Autor: Ivana Sánchez Pérez

# colores
rojo = '\033[91m'
verde = '\033[92m'
azul = '\033[94m'
magenta = '\033[95m'
amarillo = '\033[93m'
rosa = '\033[38;5;200m'
turquesa = '\033[38;5;44m'
azul_marino = '\33[38;5;67m'
lima = '\33[38;5;46m'
reset = '\033[0m'


class Libro:
    def __init__(self, titulo, autor, isbn):
        self.titulo = titulo
        self.autor = autor
        self.isbn = isbn
        self.disponible = True 

    def prestar(self):
        if self.disponible:
            self.disponible = False
            return True
        return False

    def devolver(self):
        self.disponible = True

    def __str__(self):
        estado = f"{verde}Disponible{reset}" if self.disponible else f"{rojo}Prestado{reset}"
        return f"{self.titulo} de {self.autor} (ISBN: {self.isbn}) - Estado: {estado}"

class Usuario:
    def __init__(self, nombre, id_usuario):
        self.nombre = nombre
        self.id = id_usuario
        self.libros_prestados = []

    def tomar_prestado(self, libro):
        self.libros_prestados.append(libro)
        
    def devolver_libro(self, libro): 
        if libro in self.libros_prestados:
            self.libros_prestados.remove(libro)
            return True
        return False

    def __str__(self):
        num_libros = len(self.libros_prestados)
        return (f"Usuario: {self.nombre} (ID: {self.id}) | "
                f"Libros prestados: {azul}{num_libros}{reset}")

class Biblioteca:
    def __init__(self, nombre):
        self.nombre = nombre
        self.libros = []   
        self.usuarios = []  
        print(f"{azul}Biblioteca '{nombre}' creada.{reset}")
        print(f"{azul}========================================{reset}")

    def agregar_libro(self, libro):
        self.libros.append(libro)
        print(f"{lima}Libro agregado: {libro.titulo}.{reset}")

    def registrar_usuario(self, usuario):
        self.usuarios.append(usuario)
        print(f"{lima}Usuario registrado: {usuario.nombre} (ID: {usuario.id}).{reset}")

    def _buscar_objeto_libro(self, isbn):
        for libro in self.libros:
            if libro.isbn == isbn:
                return libro
        return None

    def _buscar_objeto_usuario(self, id_usuario):
        for usuario in self.usuarios:
            if usuario.id == id_usuario:
                return usuario
        return None

    def prestar_libro(self, isbn, id_usuario):
        libro = self._buscar_objeto_libro(isbn)
        usuario = self._buscar_objeto_usuario(id_usuario)

        if not libro or not usuario:
            print(f"{rojo}¡¡ERROR!! El libro (ISBN: {isbn}) o el usuario (ID: {id_usuario}) no se han encontrado.{reset}")
            return False

        if libro.prestar(): 
            usuario.tomar_prestado(libro) 
            print(f"{verde}PRÉSTAMO REALIZADO.{reset} {usuario.nombre} tomó '{libro.titulo}'.")
            return True
        else:
            print(f"{rojo}¡¡ERROR!! El libro '{libro.titulo}' ya está prestado.{reset}")
            return False

    def devolver_libro(self, isbn, id_usuario):
        libro = self._buscar_objeto_libro(isbn)
        usuario = self._buscar_objeto_usuario(id_usuario)

        if not libro or not usuario:
            print(f"{rojo}¡¡ERROR!! El libro (ISBN: {isbn}) o el usuario (ID: {id_usuario}) no se han encontrado.{reset}")
            return False

        if usuario.devolver_libro(libro): 
            libro.devolver()
            print(f"{verde} DEVOLUCIÓN REALIZADA.{reset} {usuario.nombre} devolvió '{libro.titulo}'.")
            return True
        else:
            print(f"{rojo}¡¡ERROR!! El usuario {usuario.nombre} no tenía prestado el libro '{libro.titulo}'.{reset}")
            return False

    def libros_disponibles(self):
        disponibles = [libro.titulo for libro in self.libros if libro.disponible]
        print(f"\n{azul}LIBROS DISPONIBLES EN {self.nombre}{reset}")
        if disponibles:
            for titulo in disponibles:
                print(f"  * {amarillo}{titulo}{reset}")
            return disponibles
        print(f"{rojo}No hay libros disponibles en este momento.{reset}")
        return []

    def buscar_libro(self, titulo_buscado):
        encontrados = [libro for libro in self.libros if titulo_buscado.lower() in libro.titulo.lower()]
        
        print(f"\n{magenta}Resultados de Búsqueda para '{titulo_buscado}'{reset}")
        if encontrados:
            for libro in encontrados:
                print(f"  > {libro}")
            return encontrados
        print(f"{rojo}No se encontraron libros con el título '{titulo_buscado}'.{reset}")
        return []

# Pruebas

# Inicializar Biblioteca
biblioteca_central = Biblioteca("Ciudad Lectora")

# Crear Libros
l1 = Libro("Cien Años de Soledad", "García Márquez", "978-8437604947")
l2 = Libro("El Señor de los Anillos", "J.R.R. Tolkien", "978-0618260214")
l3 = Libro("Python para Novatos", "Varios Autores", "978-0000000001")

# Añadir Libros a la Biblioteca
biblioteca_central.agregar_libro(l1)
biblioteca_central.agregar_libro(l2)
biblioteca_central.agregar_libro(l3)

# Crear y Registrar Usuarios
u1 = Usuario("Ana Castro", 101)
u2 = Usuario("Beto Díaz", 102)
biblioteca_central.registrar_usuario(u1)
biblioteca_central.registrar_usuario(u2)

# Estado Inicial
biblioteca_central.libros_disponibles()

# Simular Préstamos
print(f"\n{azul}OPERACIONES DE PRÉSTAMO{reset}")
biblioteca_central.prestar_libro("978-8437604947", 101)
biblioteca_central.prestar_libro("978-0618260214", 102) 
biblioteca_central.prestar_libro("978-8437604947", 102) 
biblioteca_central.prestar_libro("978-0000000001", 101) 

# Consultar estados
print(f"\n{azul}ESTADO ACTUAL{reset}")
biblioteca_central.libros_disponibles()
biblioteca_central.buscar_libro("Cien Años")
print(u1)
print(u2) 

# Simular Devolución
print(f"\n{azul}OPERACIÓN DE DEVOLUCIÓN{reset}")
biblioteca_central.devolver_libro("978-0618260214", 102)

# Estado Final
print(f"\n{azul}ESTADO FINAL{reset}")
biblioteca_central.libros_disponibles() 
print(u2) 
print()
````

## 🔴 Ejercicio 8: Polimorfismo - Formas Geométricas

Crea una jerarquía polimórfica:

**Clase base `Forma`:**
- Método abstracto: `area()` (lanza NotImplementedError)
- Método abstracto: `perimetro()`

**Clases derivadas:**
- `Circulo(radio)`

- `Triangulo(base, altura, lado1, lado2, lado3)`

Cada una implementa `area()` y `perimetro()`.

Crea una función `imprimir_info(forma)` que funcione con cualquier forma (polimorfismo).

Crea una lista con diferentes formas y calcula el área total.

````python
# Título: Polimorfismos. Formas geométricas
# Autor: Ivana Sánchez Pérez

# colores
rojo = '\033[91m'
verde = '\033[92m'
azul = '\033[94m'
magenta = '\033[95m'
amarillo = '\033[93m'
rosa = '\033[38;5;200m'
turquesa = '\033[38;5;44m'
azul_marino = '\33[38;5;67m'
lima = '\33[38;5;46m'
reset = '\033[0m'

print(f"\n{azul}===== Polimorfismo - Formas geométricas ====={reset}")
print()

import math
from abc import ABC, abstractmethod

class Forma(ABC):
    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimetro(self):
        pass

# CLASES DERIVADAS

class Circulo(Forma):
    def __init__(self, radio):
        self.radio = radio

    def area(self):
        return math.pi * self.radio**2

    def perimetro(self):
        return 2 * math.pi * self.radio

class Rectangulo(Forma):
    def __init__(self, ancho, alto):
        self.ancho = ancho
        self.alto = alto

    def area(self):
        return self.ancho * self.alto

    def perimetro(self):
        return 2 * (self.ancho + self.alto)

class Triangulo(Forma):
    def __init__(self, base, altura, lado1, lado2, lado3):
        self.base = base
        self.altura = altura
        self.lado1 = lado1
        self.lado2 = lado2
        self.lado3 = lado3

    def area(self):
        return (self.base * self.altura) / 2

    def perimetro(self):
        return self.lado1 + self.lado2 + self.lado3

# FUNCIÓN POLIMÓRFICA 

def imprimir_info(forma):
    nombre_clase = type(forma).__name__
    print(f"\n{turquesa}Información de la Forma: {nombre_clase}{reset}")
    # La misma llamada (forma.area()) ejecuta código diferente en cada clase.
    print(f"{rosa}Área:{reset}{amarillo} {forma.area():.2f}{reset}")
    print(f"{rosa}Perímetro:{reset}{amarillo} {forma.perimetro():.2f}{reset}")

# PRUEBAS DE POLIMORFISMO

# Crear instancias de diferentes formas
c = Circulo(radio=5)
r = Rectangulo(ancho=10, alto=4)
t = Triangulo(base=6, altura=8, lado1=10, lado2=6, lado3=8) 

# Prueba de la función polimórfica (imprimir_info)
imprimir_info(c)
imprimir_info(r)
imprimir_info(t)

# Calcular área total de una lista de formas
print(f"\n{magenta}Cálculo del Área Total (Lista Polimórfica){reset}")

lista_formas = [c, r, t, Rectangulo(2, 2)]

area_total = 0
for forma in lista_formas:
    # Llama al método 'area()' apropiado para cada objeto de la lista
    area_total += forma.area()

print(f"{turquesa}Número total de formas:{reset}{amarillo} {len(lista_formas)}{reset}")
print(f"{turquesa}Área Total Acumulada:{reset}{amarillo} {area_total:.2f}{reset}")
````

## 🔴 Ejercicio 9: Sobrecarga de Operadores

Crea una clase `Vector2D` que represente un vector en 2D:

**Atributos:**

**Sobrecarga de operadores:**
- `__add__`: suma de vectores
- `__sub__`: resta de vectores
- `__mul__`: multiplicación por escalar
- `__eq__`: igualdad entre vectores
- `__str__`: representación "(x, y)"
- `__abs__`: magnitud del vector (√(x² + y²))

**Métodos:**
- `producto_punto(otro)`: producto punto con otro vector

Ejemplo de uso:
```python
v1 = Vector2D(3, 4)
v2 = Vector2D(1, 2)
v3 = v1 + v2  # Vector2D(4, 6)
v4 = v1 * 2   # Vector2D(6, 8)
print(abs(v1)) # 5.0
```

````python
# Título: 
# Autor: Ivana Sánchez Pérez

# colores
rojo = '\033[91m'
verde = '\033[92m'
azul = '\033[94m'
magenta = '\033[95m'
amarillo = '\033[93m'
rosa = '\033[38;5;200m'
turquesa = '\033[38;5;44m'
azul_marino = '\33[38;5;67m'
lima = '\33[38;5;46m'
reset = '\033[0m'

print(f"\n{azul}===== Sobrecarga de operadores ====={reset}")
print()

import math

class Vector2D:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    # SOBRECARGA DE OPERADORES
    
    # Suma de vectores: v1 + v2
    def __add__(self, otro):
        return Vector2D(self.x + otro.x, self.y + otro.y)

    # Resta de vectores: v1 - v2
    def __sub__(self, otro):
        return Vector2D(self.x - otro.x, self.y - otro.y)

    # Multiplicación por escalar: v1 * k (o k * v1)
    def __mul__(self, escalar):
        if isinstance(escalar, (int, float)):
            return Vector2D(self.x * escalar, self.y * escalar)
        return NotImplemented 

    # Para permitir k * v1 (multiplicación reflexiva, k va primero)
    def __rmul__(self, escalar):
        return self.__mul__(escalar)

    # Igualdad: v1 == v2
    def __eq__(self, otro):
        if isinstance(otro, Vector2D):
            return self.x == otro.x and self.y == otro.y
        return False

    # Magnitud (Valor absoluto): abs(v1)
    def __abs__(self):
        return math.sqrt(self.x**2 + self.y**2)

    # Representación legible (print(v1))
    def __str__(self):
        return f"({self.x}, {self.y})"
    
    # MÉTODOS ADICIONALES 

    def producto_punto(self, otro):
        return (self.x * otro.x) + (self.y * otro.y)

# PRUEBAS DE LA CLASE Y OPERADORES 

print(f"{rosa}Creación de Vectores{reset}")
v1 = Vector2D(3, 4)
v2 = Vector2D(1, 2)
v_cero = Vector2D(0, 0)

print(f"{turquesa}v1:{reset}{amarillo} {v1}{reset}")
print(f"{turquesa}v2:{reset}{amarillo} {v2}{reset}")

# Prueba de Suma y Resta (Sobrecarga de __add__ y __sub__)
print(f"\n{rosa}Operaciones Vectoriales (Sobrecarga){reset}")
v3 = v1 + v2
print(f"{turquesa}v1 + v2:{reset}{amarillo} {v3}{reset}")   

v_resta = v1 - v2
print(f"{turquesa}v1 - v2:{reset}{amarillo} {v_resta}{reset}") 

# Prueba de Multiplicación por Escalar (Sobrecarga de __mul__ y __rmul__)
k = 2
v4 = v1 * k
print(f"{turquesa}v1 * {k}:{reset}{amarillo} {v4}{reset}")     

v5 = k * v1
print(f"{turquesa}{k} * v1:{reset}{amarillo} {v5}{reset}")    

# Prueba de Igualdad (Sobrecarga de __eq__)
v6 = Vector2D(3, 4)
print(f"\n{rosa}Operaciones de Comparación{reset}")
print(f"{turquesa}v1 == v6:{reset}{amarillo} {v1 == v6}{reset}") 
print(f"{turquesa}v1 == v2:{reset}{amarillo} {v1 == v2}{reset}") 

# Prueba de Magnitud (Sobrecarga de __abs__)
print(f"\n{rosa}Funciones Integradas (Sobrecarga){reset}")
magnitud_v1 = abs(v1)
print(f"{turquesa}Magnitud (abs(v1)):{reset}{amarillo} {magnitud_v1:.1f}{reset}") 

# Prueba de Producto Punto (Método normal)
dot_product = v1.producto_punto(v2)
print(f"{turquesa}Producto Punto (v1 . v2):{reset}{amarillo} {dot_product}{reset}") 
````

## 🔴 Ejercicio 10: Sistema de Empleados con Herencia

Crea una jerarquía de clases para empleados:

**Clase base `Empleado`:**
- Atributos: `nombre`, `id`, `salario_base`
- Método: `calcular_salario()` (abstracto)
- Método: `__str__()`

**Clases derivadas:**

**`EmpleadoTiempoCompleto(Empleado)`:**
- Atributo: `bono_anual`
- `calcular_salario()`: salario_base + bono_anual/12

**`EmpleadoPorHoras(Empleado)`:**
- Atributos: `horas_trabajadas`, `tarifa_hora`
- `calcular_salario()`: horas × tarifa (horas extra ×1.5 si >40)

**`EmpleadoComision(Empleado)`:**
- Atributos: `ventas`, `porcentaje_comision`
- `calcular_salario()`: salario_base + (ventas × comision)

**Clase `Empresa`:**
- Atributos: `nombre`, `empleados` (lista)
- Métodos:
  - `contratar(empleado)`
  - `despedir(id_empleado)`
  - `calcular_nomina_total()`
  - `empleado_mejor_pagado()`

Crea una empresa con diferentes tipos de empleados y calcula la nómina total.

````python
# Título: Sistema de empleados con herencia
# Autor: Ivana Sánchez Pérez

# colores
rojo = '\033[91m'
verde = '\033[92m'
azul = '\033[94m'
magenta = '\033[95m'
amarillo = '\033[93m'
rosa = '\033[38;5;200m'
turquesa = '\033[38;5;44m'
azul_marino = '\33[38;5;67m'
lima = '\33[38;5;46m'
reset = '\033[0m'

print(f"\n{rosa}===== Sistema de nóminas de PetTracer ====={reset}")
print()

import math
from abc import ABC, abstractmethod

# Clase base abstracta
class Empleado(ABC):
    def __init__(self, nombre, id_empleado, salario_base):
        self.nombre = nombre
        self.id = id_empleado
        self.salario_base = salario_base
        
    @abstractmethod
    def calcular_salario(self):
        pass

    def __str__(self):
        return (f"{self.nombre} (ID: {self.id}, Tipo: {type(self).__name__}) | "
                f"Sueldo Mensual: {amarillo}{self.calcular_salario():.2f}€{reset}")

# CLASES DERIVADAS (Polimorfismo en calcular_salario)

class EmpleadoTiempoCompleto(Empleado):
    def __init__(self, nombre, id_empleado, salario_base, bono_anual):
        super().__init__(nombre, id_empleado, salario_base)
        self.bono_anual = bono_anual

    def calcular_salario(self):
        return self.salario_base + (self.bono_anual / 12)

class EmpleadoPorHoras(Empleado):
    def __init__(self, nombre, id_empleado, salario_base, horas_trabajadas, tarifa_hora):
        super().__init__(nombre, id_empleado, salario_base)
        self.horas_trabajadas = horas_trabajadas
        self.tarifa_hora = tarifa_hora
    
    def calcular_salario(self):
        horas = self.horas_trabajadas
        tarifa = self.tarifa_hora
        
        if horas > 40:
            horas_normales = 40
            horas_extra = horas - 40
            pago_horas = (horas_normales * tarifa) + (horas_extra * tarifa * 1.5)
        else:
            pago_horas = horas * tarifa
            
        return pago_horas
        
class EmpleadoComision(Empleado):
    def __init__(self, nombre, id_empleado, salario_base, ventas, porcentaje_comision):
        super().__init__(nombre, id_empleado, salario_base)
        self.ventas = ventas
        self.porcentaje_comision = porcentaje_comision 

    def calcular_salario(self):
        comision = self.ventas * self.porcentaje_comision
        return self.salario_base + comision

# CLASE CONTROLADORA 

class Empresa:
    def __init__(self, nombre):
        self.nombre = nombre
        self.empleados = []
        
    def contratar(self, empleado):
        self.empleados.append(empleado)
        print(f"{lima}Contratado: {empleado.nombre} como {type(empleado).__name__}.{reset}")
        
    def despedir(self, id_empleado):
        empleado_a_despedir = None
        for emp in self.empleados:
            if emp.id == id_empleado:
                empleado_a_despedir = emp
                break
        
        if empleado_a_despedir:
            self.empleados.remove(empleado_a_despedir)
            print(f"{rojo}Despedido: {empleado_a_despedir.nombre} (ID: {id_empleado}).{reset}")
            return True
        print(f"{rojo}¡¡ERROR!! EL empleado con ID {id_empleado} no se ha encontrado.{reset}")
        return False

    def calcular_nomina_total(self):
        nomina_total = 0
        print(f"\n{rosa}Cálculo de Nómina Mensual de {self.nombre}{reset}")
        for emp in self.empleados:
            salario = emp.calcular_salario() 
            nomina_total += salario
            print(f"{turquesa}[ID {emp.id}] {emp.nombre:<30}: {amarillo}{salario:.2f}€{reset}")
            
        return nomina_total

    def empleado_mejor_pagado(self):
        if not self.empleados:
            return None
            
        mejor_pagado = max(self.empleados, key=lambda emp: emp.calcular_salario())
        return mejor_pagado

# CREACIÓN Y SIMULACIÓN DE PETTRACER 

pettracer = Empresa("PetTracer")

empleados_a_contratar = [
    EmpleadoTiempoCompleto("Ivana Sánchez Pérez", 101, salario_base=5000.00, bono_anual=12000.00),
    EmpleadoTiempoCompleto("Raquel Carmona Jiménez", 102, salario_base=1800.00, bono_anual=2400.00),
    EmpleadoComision("Francisco González Romero", 103, salario_base=1200.00, ventas=25000.00, porcentaje_comision=0.08),
    EmpleadoComision("Inmaculada Rodríguez Asencio", 104, salario_base=1200.00, ventas=15000.00, porcentaje_comision=0.10),
    EmpleadoTiempoCompleto("David Suarez Ruiz", 105, salario_base=1600.00, bono_anual=1800.00),
    EmpleadoPorHoras("Gustavo Pastor Trillo", 106, salario_base=0.00, horas_trabajadas=45, tarifa_hora=12.00)
]

# Contratar empleados
for emp in empleados_a_contratar:
    pettracer.contratar(emp)

# Calcular Nómina Total (Demuestra el Polimorfismo)
nomina = pettracer.calcular_nomina_total()
print(f"\n{verde}>>> NÓMINA TOTAL DE {pettracer.nombre}:{reset}{amarillo} {nomina:.2f}€{reset}")

# Encontrar el mejor pagado
mejor = pettracer.empleado_mejor_pagado()
print(f"\n{rosa}Información Adicional{reset}")
if mejor:
    print(f"{turquesa}Empleado mejor pagado:{reset} {mejor.nombre} con un sueldo de {amarillo}{mejor.calcular_salario():.2f}€{reset}")

# Ejemplo de Despido
pettracer.despedir(106)
print(f"\n{turquesa}Lista de empleados después del despido:{reset}")
for emp in pettracer.empleados:
    print(f"* {emp.nombre}")

# Recalcular Nómina
nueva_nomina = pettracer.calcular_nomina_total()
print(f"\n{verde}>>> NUEVA NÓMINA TOTAL DE {pettracer.nombre}:{reset}{amarillo} {nueva_nomina:.2f}€{reset}")
````

````python
# Título: Decoradores y Context Manager
# Autor: Ivana Sánchez Pérez

# colores
rojo = '\033[91m'
verde = '\033[92m'
azul = '\033[94m'
magenta = '\033[95m'
amarillo = '\033[93m'
rosa = '\033[38;5;200m'
turquesa = '\033[38;5;44m'
azul_marino = '\33[38;5;67m'
lima = '\33[38;5;46m'
reset = '\033[0m'

print(f"\n{rosa}===== Decoradores y Context Manager ====={reset}")

import time

class Cronometro:
    def __init__(self, operacion):
        self.operacion = operacion
        self.tiempo_inicio = None

    def __enter__(self):
        self.tiempo_inicio = time.time()
        print(f"{turquesa}Iniciando cronómetro para:{reset}{amarillo} {self.operacion}{reset}")
        return self

    def __exit__(self, exc_type, exc_val, exc_tb):
        tiempo_fin = time.time()
        duracion = tiempo_fin - self.tiempo_inicio
        
        # Muestra el resultado
        print(f"\n{magenta}Bloque {self.operacion} finalizado.{reset}")
        print(f"{resultado}Resultado:{reset}{amarillo} '{self.operacion}' tardó {duracion:.4f} segundos.{reset}")
        
        return False 

# Prueba 

print(f"\n{magenta}Prueba de Cronometro (Context Manager) {reset}")

with Cronometro("Cálculo de suma grande"):
    suma = sum(range(1000000))
    print(f"{turquesa}Suma calculada hasta{reset}{amarillo} {suma}.{reset}")

with Cronometro("Operación corta"):
    resultado = 1 + 1
    time.sleep(0.005)
    print(f"{turquesa}Resultado de la operación:{reset}{amarillo} {resultado}{reset}")
````
