## 🟢 Ejercicio 1: El Conversor de Temperatura

Crea un programa que convierta una temperatura de Celsius a Fahrenheit y Kelvin.

**Fórmulas:**
- Fahrenheit = (Celsius × 9/5) + 32
- Kelvin = Celsius + 273.15

Usa la variable `temperatura_celsius = 25` y muestra los tres valores.

````python
# Título: El Conversor de Temperatura
# Autor: Ivana Sánchez Pérez

# Colores
rojo = '\033[91m'
verde = '\033[92m'
azul = '\033[94m'
magenta = '\033[95m'
amarillo = '\033[93m'
rosa = '\033[38;5;200m' 
reset = '\033[0m'  

# Variable de temperatura en Celsius
temperatura_celsius = 25

# Conversión a Fahrenheit: F = (C × 9/5) + 32
temperatura_fahrenheit = (temperatura_celsius * 9/5) + 32

# Conversión a Kelvin: K = C + 273.15
temperatura_kelvin = temperatura_celsius + 273.15

print()  
print(f"{azul}Temperatura en Celsius: {temperatura_celsius}°C{reset}")
print(f"{rojo}Temperatura en Fahrenheit: {temperatura_fahrenheit}°F{reset}")
print(f"{verde}Temperatura en Kelvin: {temperatura_kelvin}K{reset}")
print()  
````

## 🟡 Ejercicio 6: El Precio Final

Una tienda online vende un producto por 45.99€. Aplica:
- Un descuento del 15%
- Un IVA del 21%

Calcula el precio final y muéstralo con 2 decimales.

**Pista:** Usa `round()` o formateo de strings.

````python
# Título: El precio Final
# Autor: Ivana Sánchez Pérez

# colores
rojo = '\033[91m'
verde = '\033[92m'
azul = '\033[94m'
magenta = '\033[95m'
amarillo = '\033[93m'
rosa = '\033[38;5;200m' 
reset = '\033[0m'

print(f"\n{azul}=== El Precio Final ==={reset}")
print()
# Precio Base
precio_base = 45.99
print(f"{rojo}Precio base: {precio_base}{reset}")

# Cálculos
descuento = precio_base * 0.15
precio_con_descuento = precio_base - descuento
iva = precio_con_descuento * 0.21
precio_final = precio_con_descuento + iva

print(f"\n{verde}=== CÁLCULOS ==={reset}")
print(f"{magenta}Descuento (15%): {descuento:.2f}€{reset}")
print(f"{magenta}Precio con descuento: {precio_con_descuento:.2f}€{reset}")
print(f"{magenta}IVA (21%): {iva:.2f}€{reset}")

print(f"\n{amarillo}=== RESULTADO FINAL ==={reset}")
print(f"{magenta}Precio final: {precio_final:.2f}€{reset}")
print()




````

## 🔴 Ejercicio 7: El Puzzle de las Divisiones

Dado el número 100, realiza las siguientes operaciones y explica la diferencia:

1. `100 / 3` (división normal)
2. `100 // 3` (división entera)
3. `100 % 3` (módulo/resto)
4. Verifica que: `(100 // 3) * 3 + (100 % 3) == 100`

Muestra cada resultado y añade comentarios explicando qué hace cada operador.

````python
# Título: El Puzzle de las divisiones
# Autor: Ivana Sánchez Pérez

# colores
rojo = '\033[91m'
verde = '\033[92m'
azul = '\033[94m'
magenta = '\033[95m'
amarillo = '\033[93m'
rosa = '\033[38;5;200m' 
reset = '\033[0m'

print(f"\n{azul}=== El puzzle de las Divisiones ==={reset}")
print()

# Variables
numero = 100
divisor = 3

print(f"{verde}Número: {numero}{reset}")
print(f"{verde}Divisor: {divisor}{reset}")

# División normal
print(f"\n{rojo}1. 100 / 3 (división normal): {numero / divisor}{reset}")

# División entera 
print(f"{amarillo}2. 100 // 3 (división entera): {numero // divisor}{reset}")

# 3 Módulo/Resto
print(f"{magenta}3. 100 % 3 (módulo/resto): {numero % divisor}{reset}")

# 4 Verificación
cociente_entero = numero // divisor
resto = numero % divisor
verificacion = (cociente_entero * divisor) + resto

print(f"\n{verde}4. Verificación: ({cociente_entero} * {divisor}) + {resto} == {verificacion}{reset}")

print(f"{rosa}El resultado es igual a 100? {verificacion == numero}{reset}")
print()
````

## 🔴 Ejercicio 8: Calculadora de IMC (Índice de Masa Corporal)

Crea un programa que calcule el IMC usando la fórmula:

**IMC = peso (kg) / altura² (m)**

Usa estas variables:
- `peso = 70` kg
- `altura = 175` cm (¡cuidado! debes convertirla a metros)

Muestra el resultado con un mensaje claro.

````python
# Título: Calculadora de IMC
# Autor: Ivana Sánchez Pérez

# colores
rojo = '\033[91m'
verde = '\033[92m'
azul = '\033[94m'
magenta = '\033[95m'
amarillo = '\033[93m'
rosa = '\033[38;5;200m' 
reset = '\033[0m'

print(f"\n{azul}=== Calculadora de IMC ==={reset}")
print()

# Variables
peso = 70 # kg
altura_cm = 175

print(f"{amarillo}Datos Iniciales{reset}")
print(f"{amarillo}---------------{reset}")
print(f"{azul}Peso: {peso} kg{reset}")
print(f"{azul}Altura: {altura_cm} cm{reset}")

# Conversión de la altura a metros
altura_m = altura_cm /100
print(f"\n{amarillo}Conversión{reset}")
print(f"{amarillo}----------{reset}")
print(f"{azul}Altura en metros: {altura_cm} / 100 = {altura_m}{reset}")

# Cálcula del Índice de Masa Muscular
imc = peso / (altura_m ** 2)
print(f"\n{amarillo}Cálculo del IMC{reset}")
print(f"{amarillo}---------------{reset}")
print(f"{azul}Fórmula: IMC = peso (kg) / altura **  (m){reset}")
print(f"{azul}IMC = {peso} / ({altura_m}) ** {reset}")
print(f"{azul}IMC = {peso} / {altura_m:.4f} ** {reset}")
print(f"{azul}IMC = {imc:.2f}{reset}")

print(f"\n{verde}RESULTADO: Tu IMC es de {imc:.2f}{reset}")
print()

````

Crea un nombre de usuario a partir de:

## 🎯 Ejercicio Bonus: El Código Secreto

Descifra este código:

```python
x = 5
y = x + 3
z = y * 2
resultado = z ** 2 - x * y
```

Sin ejecutar el código, **calcula mentalmente** qué valor tendrá `resultado`.

Luego, ejecuta el código para verificar tu respuesta. ¿Acertaste?

````python
# Título: El Código Secreto
# Autor: Ivana Sánchez Pérez

# colores
rojo = '\033[91m'
verde = '\033[92m'
azul = '\033[94m'
magenta = '\033[95m'
amarillo = '\033[93m'
rosa = '\033[38;5;200m' 
reset = '\033[0m'

print(f"\n{azul}=== El Código Secreto ==={reset}")
print()

# Código
x = 5
y = x + 3
z = y * 2
resultado = z **2 - x * y

print(f"{amarillo}Cálculo desglosado{reset}")
print(f"{amarillo}------------------{reset}")
print(f"\n{rosa}x = {x}{reset}")
print(f"{rosa}y = x + 3 = {x} + 3 = {y}{reset}")
print(f"{rosa}z = Y × 2 = {y} × 2 = {z}{reset}")
print(f"{rosa}resultado = z² - x × y = {z}² - {x} × {y}{reset}")

# Cálculo final
print(f"\n{amarillo}Cálculo final{reset}")
print(f"{amarillo}-------------{reset}")
print(f"{rosa}{z}² = {z ** 2}{reset}")
print(f"{rosa}{x} x {y} = {x * y}{reset}")
print(f"{rosa}{z ** 2} - {x * y} = {resultado}{reset}")

# Resultado final
print(f"\n{verde}¡El resultado es: {resultado}!{reset}{reset}")
print()
print(f"{azul} Mi cálculo mental ha sido correcto porque el orden de las operaciones ya está definido; primero la potencia, luego la multiplicación y división, y por último la suma y la resta. Por eso no hacen falta paréntesis.{reset}")
print()



````
